```python
# pandas: build a small frame, then describe + group it.
df = pd.DataFrame({
    'group': np.repeat(['a', 'b'], 50),
    'val': np.linspace(0, 1, 100),
})
print(df.groupby('group')['val'].agg(['count', 'mean', 'std']))
```
