## Cleaning Strategy and Assumptions

The preprocessing workflow uses three main steps:

1. Columns with more than 50% missing values are removed because they contain too little reliable information. In this dataset, `extra_data` is removed for this reason.

2. Missing values in the numeric columns `age`, `income`, and `score` are filled using the median. Median imputation is used because it is less sensitive to extreme values than the mean.

3. The numeric columns `age`, `income`, and `score` are normalized using Min-Max scaling so that their values fall between 0 and 1.

The `zipcode` and `city` columns are kept unchanged because they are identifier/categorical fields and should not be normalized.

This cleaning strategy assumes that the missing numeric values can reasonably be represented by the median of the observed values and that removing a column with more than 50% missing data is preferable to imputing most of its values.
