```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print("Original array:", arr)
print("Add 10:", arr + 10)
print("Multiply by 2:", arr * 2)
print("Squared:", arr ** 2)
```

```python
import time

large_array = np.arange(1_000_000)

# Loop execution
start = time.perf_counter()

loop_result = []
for x in large_array:
    loop_result.append(x * 2)

loop_time = time.perf_counter() - start


# Vectorized execution
start = time.perf_counter()

vectorized_result = large_array * 2

vectorized_time = time.perf_counter() - start


print(f"Loop time: {loop_time:.6f} seconds")
print(f"Vectorized time: {vectorized_time:.6f} seconds")
print(f"Vectorized execution is approximately {loop_time / vectorized_time:.1f}x faster")
```

```python
from pathlib import Path
import pandas as pd

# Locate project root
ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

DATA_PATH = ROOT / "data" / "raw" / "starter_data.csv"

print("Data path:", DATA_PATH)

df = pd.read_csv(DATA_PATH)
```

```python
summary_stats = df.describe()

summary_stats
```

```python
grouped_summary = (
    df.groupby("category")["value"]
    .agg(["count", "mean", "sum", "min", "max"])
    .reset_index()
)

grouped_summary
```

```python
PROCESSED_PATH = ROOT / "data" / "processed"
PROCESSED_PATH.mkdir(parents=True, exist_ok=True)

summary_stats.to_csv(PROCESSED_PATH / "summary.csv")

print("Saved to:", PROCESSED_PATH / "summary.csv")
```

```python
grouped_summary.to_csv(
    PROCESSED_PATH / "grouped_summary.csv",
    index=False
)

print("Groupby summary saved.")
```
